```julia
using Plots
using DelimitedFiles
signal_1 = readdlm("D:/Documents/Study/3 курс/оснЦОС/signal_1.csv", ',', Float64)
signal_2 = readdlm("D:/Documents/Study/3 курс/оснЦОС/signal_2.csv", ',', Float64)
x_s1 = vec(signal_1)
x_s2 = vec(signal_2)

p1 = scatter(x_s1, title = "Сигнал 1 дискретный", xlabel = "Индекс n", ylabel = "Амплитуда",label = "x1[n]",
linewidth=2, markersize = 1,markerstrokewidth = 0, color=:violetred1)

p2 = scatter(x_s2,title = "Сигнал 2 дискретный",xlabel = "Индекс n",ylabel = "Амплитуда",label = "x2[n]",
linewidth=2, markersize = 1,markerstrokewidth = 0, color=:chocolate2)


display(p1)
display(p2)
savefig(p1, "signal_1.png")
savefig(p2, "signal_2.png")
```

```julia
function correlation(x, h)
    N = length(x)
    M = length(h)
    len = N + M - 1
    r = zeros(len)
    for m in 1:len
        lag = m - M
        for n in 1:N
            j = n - lag
            if 1 <= j <= M
                r[m] += x[n]*h[j]
            end
        end
    end
    
    return r
end
N = length(x_s1)
M = length(x_s2)

lags = -(M-1):(N-1)
corl_s1s2 = correlation(x_s1,x_s2)
p_s1s2 = scatter(lags, corl_s1s2, title = "Коррлеряция x1[n] ⋆ x2[n]", xlabel = "Сдвиг",
ylabel = "Амплитуда",label = "r[m]", linewidth=2, markersize = 1,
markerstrokewidth = 0, color=:turquoise1)
savefig(p_s1s2, "s1s2.png")
display(p_s1s2)
```

```julia

max_value, max_index = findmax(corl_s1s2)
println("Первый пик:")
println("Сдвиг: ", lags[max_index])
println("Значение: ", max_value)
#убираем первый пик
r_copy = copy(corl_s1s2)
r_copy[max(1, max_index-5000):min(end, max_index+5000)] .= -Inf

#находим второй пик
second_value, second_index = findmax(r_copy)

println("Второй пик:")
println("Сдвиг: ", lags[second_index])
println("Значение: ", second_value)
```
